```python
from theia.radar_equation import radar_eq_max_dist
from theia.types import Point, Polarization, Radar
from theia.coverage import calculate_coverage
```

# Config

```python
radar = Radar(
    id=585,
    point=Point(
        lat=47.36700085728634,
        lon=8.537724304199216,
        alt=407.83600886023686,
    ),
    power=20000,
    erp=800,
    antenna_height=10.0,
    diameter=2.0,
    frequency=1000.0,
    pulse_width=1,
    cpi_pulses=1,
    bandwidth=1,
    pfa=1e-6,
    min_elevation=-20.0,
    max_elevation=60.0,
    rotation_time=10.0,
    polarization=Polarization.HORIZONTAL,
)

target_cross_section = 2.0
target_flight_height = 1000.0
```

# Calculate coverage

```python
max_dist = radar_eq_max_dist(
    radar,
    target_cross_section,
)

coverage = calculate_coverage(radar.point, max_dist, target_flight_height, d_theta=1)
```

# Plot coverage

```python
import folium


map = folium.Map(location=(radar.lat, radar.lon))
# Add  marker for the radar site.
folium.Marker(
    location=(radar.lat, radar.lon),
    icon=folium.Icon(prefix="fa", icon="satellite-dish"),
    tooltip=f"My radar<br />Lat: {radar.lat:.4f}°<br />Lon: {radar.lon:.4f}°<br />Alt [MASL]: {radar.alt:.1f}",
).add_to(map)
# Add an overlay for the coverage.
folium.GeoJson(coverage).add_to(map)
# Add a popup to display (lat, lon) on click.
folium.LatLngPopup().add_to(map)

map.save("map.html")

map
```
