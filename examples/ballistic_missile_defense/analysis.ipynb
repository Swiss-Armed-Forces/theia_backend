```python
import numpy as np
import pandas as pd
import seaborn as sns

from theia.simulation.theia_logging import LogLoader


loader = LogLoader("../../test.json")

t_start = loader._snapshots[0].time
t_start = t_start.timestamp()

t_stop = loader._snapshots[-1].time
t_stop = t_stop.timestamp()

detections = loader.blue_monostatic_radar_detections
df_detections = pd.DataFrame(
    [
        {
            "id": d.detection_id,
            "time": d.time.timestamp() - t_start,
            "target_id": d.target.id,
            "sensor_id": d.radar.id,
            "elevation": np.rad2deg(d.elevation_angle),
        }
        for d in detections
    ]
)
df_detections["Sensor Name"] = ""
i = df_detections.query("`sensor_id` == 0").index
df_detections.loc[i, "Sensor Name"] = "Pilatus"
i = df_detections.query("`sensor_id` == 1").index
df_detections.loc[i, "Sensor Name"] = "Scopi"
i = df_detections.query("`sensor_id` == 2").index
df_detections.loc[i, "Sensor Name"] = "Weissflue"
i = df_detections.query("`sensor_id` == 3").index
df_detections.loc[i, "Sensor Name"] = "Weisshorn"
df_detections
```

```python
from matplotlib import pyplot as plt


fig, ax = plt.subplots(figsize=(8, 4.5))

sns.scatterplot(data=df_detections, x="time", y="elevation", hue="Sensor Name", ax=ax)
ax.set_xlim([0, ax.get_xlim()[1]])
ax.set_ylim([0, 50])
ax.set_ylabel("Detection elevation [°]", fontsize=20)
ax.set_xlabel("Time [s] since sim start", fontsize=20)
```

```python
t_reaction = t_stop - t_start - df_detections.iloc[0, :]["time"]
print(f"Reaction time from first detection to impact: {t_reaction:.0f} s")
```
